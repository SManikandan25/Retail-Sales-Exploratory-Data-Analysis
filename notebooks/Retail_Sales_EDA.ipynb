**Step 4: Handling Missing Values**

```python
df.isnull().sum()
```

```python
df[df['Postal Code'].isnull()]
```

**Observation:** Only 11 records contained missing values in the Postal Code column. Since Postal Code is an identifier field, missing values were investigated based on City and State information before applying any imputation strategy.​‌


```python
df[df['City'] == 'Burlington'][['City','State','Postal Code']]
```

```python
df.loc[
    (df['City'] == 'Burlington') &
    (df['State'] == 'Vermont') &
    (df['Postal Code'].isna()),
    'Postal Code'
] = 5401
```

```python
df["Postal Code"].isnull().sum()
```

**Observation:** The Postal Code column contained 11 missing values. Upon further investigation, all missing records belonged to the same City (Burlington) and State (Vermont).

**Reasoning:** Since Postal Code is a location identifier and the City and State information were already available, the missing values could be accurately identified. Removing these records would result in unnecessary data loss.

**Action Taken:** The missing Postal Code values were replaced with the valid Burlington, Vermont postal code (05401).

**Result:** All missing values in the Postal Code column were successfully resolved while preserving the original transaction records.

**Step 5: Data Type Verification** 

```python
df.info()
```

```python
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    dayfirst=True 
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    dayfirst=True
)
```

The Order Date and Ship Date columns were converted from string to datetime format. The dayfirst=True parameter was used because the dates were stored in DD-MM-YYYY format,ensuring accurate date conversion for further analysis.​‌


```python
df[['Order Date','Ship Date']].dtypes
```

**Observation:** The Order Date and Ship Date columns were stored as string data types. 
                 Both columns were converted to datetime format to support time-based analysis such as yearly, monthly, and daily sales trends.
